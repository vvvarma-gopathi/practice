1.Write Python code using MongoClient to connect to a local MongoDB instance, select schoolDB and students,
and verify the connection with a ping.

2.Insert one student document using insert_one() with name, age, department, marks, skills, city, and a nested
fees object.

```python
from pymongo import MongoClient
client=MongoClient('mongodb://localhost:27017')
db=client['school']
collection=db['students']
collection.insert_one({'id':100,'name':'varma','age':23,'dept':"CSE",'marks':76,'fees':{'total':40000,'paid':18500}})
print("record successfully saved into mongo..")
```

3.Print the names and marks of all students whose marks are greater than or equal to 80. Remember that find()
returns a cursor that must be iterated.

```python
from pymongo import MongoClient
client=MongoClient('mongodb://localhost:27017')
db=client['school']
collection=db['students']
cursor=collection.find({'marks':{'$gt':80}},{'_id':0,'name':1,'marks':1,'city':1})
print('-'*50)
for record in cursor:
    print(record)
```

4.Display only name, department, and marks for the top 3 scorers, sorted by marks in descending order.

```python
from pymongo import MongoClient
client = MongoClient('mongodb://localhost:27017')
db=client['school']
collection=db['students']
cursor=collection.find({},{'_id':0,'name':1,'age':1,'marks':1}).sort('marks',-1).limit(3)
print("-"*50)
for record in cursor:
    print(record)
```

5.Add 5 marks to every CSE student using update_many() and $inc. Display the modified count.

```python
from pymongo import MongoClient
client=MongoClient("mongodb://localhost:27017")
db=client['school']
collection=db['students']
collection.update_many({'dept':"CSE"},{'$inc':{'marks':5}})
print("marks 5 incremented for cse students...")
```

6.Add the skill git to every CSE student's skills without creating duplicates. Use the appropriate array update
operator.

```python
from pymongo import MongoClient
client = MongoClient('mongodb://localhost:27017')
db=client['school']
students=db['students']
students.update_many({'dept':"CSE"},{'$addToSet':{'skills':'git'}})
print("Successfully updated skills to CSE students")
```

7.Write an update_one() operation that inserts a student if that student does not already exist, using upsert=True.

```python
from pymongo import MongoClient
client=MongoClient("mongodb://localhost:27017")
db=client['school']
students=db['students']
result=students.update_one({'name':'vennila'},{'$set':{'id':30,'dept':'MECH','marks':60,'skills':['python','autocad','shading']}},upsert=True)
print(result.modified_count)
```

8.Create an aggregation pipeline to show average marks and number of students for each department, sorted by
average marks from highest to lowest.

```python
from pymongo import MongoClient
client=MongoClient("mongodb://localhost:27017")
db=client['school']
print(db)
students=db['students']
print(students)
results = students.aggregate([{"$group":{"_id":"$dept",'average_marks':{'$avg':'$marks'},'count':{'$sum':1}}},{'$sort':{'average_marks':-1}}])
print("-"*50)
for record in results:
    print(record)
```

9.Using $lookup, join students with departments and display each student's name, department code, and HOD.

```python
from pymongo import MongoClient
client=MongoClient("mongodb://localhost:27017")
db=client['school']
students=db['students']
results=students.aggregate([{'$lookup':{'from':'departments','localField':'dept','foreignField':'dept_name','as':'dept_info'}}
                            ,{'$unwind':'$dept_info'},
                            {'$project':{'name':1,'_id':0,'department':'$dept','Hod':'$dept_info.Hod'}},
                            {'$sort':{'department':1}}])
print("-"*80)
for record in results:
    print(record)
```
