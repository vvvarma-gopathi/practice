15.Use pd.read_sql() to join students, enrollments, and courses and display student name with course name.

```python
import pandas as pd
import psycopg2
conn=psycopg2.connect(
    host="localhost",
    database='school',
    user='postgres',
    password='varma66',
    port='5432'
)
query = """
SELECT s.name AS student_name, c.course_name
FROM students s
JOIN enrollments e ON s.student_id = e.student_id
JOIN courses c ON e.course_id = c.course_id
"""
df=pd.read_sql(query,conn)
print("-"*60)
print(df)
```

16.Use a subquery to display students whose marks are above the average marks of all students.

```python
import psycopg2
conn=psycopg2.connect(
    host="localhost",
    database='school',
    user='postgres',
    password='varma66',
    port='5432'
)
cur=conn.cursor()
cur.execute("select name,marks,course_name from students where marks>(select avg(marks) from students)")
results=cur.fetchall()
print("-"*60)
for record in results:
    print(record)
```

17.Create add_student(name, marks) that inserts a student, gets the new id using RETURNING id, and commits the
transaction.

```python
import psycopg2
conn=psycopg2.connect(
    host="localhost",
    database='school',
    user='postgres',
    password='varma66',
    port='5432'
)
def add_student(id,name,marks,course_name):
    cur=conn.cursor()
    cur.execute("insert into students(id,name,marks,course_name) values (%s,%s,%s,%s) returning id",(id,name,marks,course_name))
    student_id=cur.fetchone()[0]
    conn.commit()
    cur.close()
    return student_id

id=add_student(7,'jasmine',87,'mern')
print("new id is :",id)
```

18.Write a try/except example around a database operation. On failure, call conn.rollback() so the connection can
continue to be used.

```python
import psycopg2
conn=psycopg2.connect(
    host="localhost",
    database='school',
    user='postgres',
    password='varma66',
    port='5432'
)
cur=conn.cursor()
try:
    cur.execute("insert into students(name,marks) values ('vardhan',98)")
    conn.commit()
    print("Data successfully saved..")
except Exception as e:
    conn.rollback()
    print("Error occured rollbacked: ",e)
```
